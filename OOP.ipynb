```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand  # Attribute
        self.model = model  # Attribute

    def display_info(self):  # Method
        return f"{self.brand} {self.model}"

# Creating an Object (Instance)
car1 = Car("Toyota", "Camry")
print(car1.display_info())  # Output: Toyota Camry
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private Attribute

    def get_balance(self):  # Getter
        return self.__balance

    def deposit(self, amount):  # Setter
        if amount > 0:
            self.__balance += amount

# Using Encapsulation
account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())  # Output: 1500
```

```python
class A:
    def method_a(self):
        return "Method A"

class B:
    def method_b(self):
        return "Method B"

class C(A, B):  # Multiple Inheritance
    pass

obj = C()
print(obj.method_a())  # Output: Method A
print(obj.method_b())  # Output: Method B
```

```python
class Bird:
    def fly(self):
        return "Birds can fly"

class Penguin(Bird):
    def fly(self):
        return "Penguins cannot fly"

bird = Bird()
penguin = Penguin()

print(bird.fly())      # Output: Birds can fly
print(penguin.fly())   # Output: Penguins cannot fly
```

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):  # String representation
        return f"Book: {self.title}"

    def __len__(self):  # Define behavior for len()
        return self.pages

book = Book("Python Basics", 300)
print(str(book))  # Output: Book: Python Basics
print(len(book))  # Output: 300
```

```python
# LIBRARY MANAGEMENT SYSTEM
from datetime import datetime

# Base class
class Person:
    def __init__(self, name, email):
        self.name = name
        self.email = email

# Member class
class Member(Person):
    def __init__(self, name, email, member_id):
        super().__init__(name, email)
        self.member_id = member_id
        self.borrowed_books = []

    def borrow_book(self, book, library):
        if library.lend_book(book, self):
            self.borrowed_books.append(book)
            print(f"{self.name} borrowed '{book.title}'")
        else:
            print(f"Sorry, '{book.title}' is not available.")

    def return_book(self, book, library):
        if book in self.borrowed_books:
            library.receive_book(book, self)
            self.borrowed_books.remove(book)
            print(f"{self.name} returned '{book.title}'")
        else:
            print(f"{self.name} does not have '{book.title}' borrowed.")

# Librarian class
class Librarian(Person):
    def __init__(self, name, email, employee_id):
        super().__init__(name, email)
        self.employee_id = employee_id

    def add_book(self, book, library):
        library.add_book(book)
        print(f"{self.name} added '{book.title}' to the library.")

# Book class
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_borrowed = False

# Library class
class Library:
    def __init__(self, name):
        self.name = name
        self.books = []
        self.history = []

    def add_book(self, book):
        self.books.append(book)

    def lend_book(self, book, member):
        if book in self.books and not book.is_borrowed:
            book.is_borrowed = True
            self.history.append((book.title, member.name, 'borrowed', datetime.now()))
            return True
        return False

    def receive_book(self, book, member):
        if book in self.books:
            book.is_borrowed = False
            self.history.append((book.title, member.name, 'returned', datetime.now()))
            return True
        return False

    def display_books(self):
        print(f"\nBooks in {self.name} Library:")
        for book in self.books:
            status = "Borrowed" if book.is_borrowed else "Available"
            print(f"  - {book.title} by {book.author} [{status}]")

    def show_history(self):
        print("\nTransaction History:")
        for record in self.history:
            title, user, action, timestamp = record
            print(f"{timestamp.strftime('%Y-%m-%d %H:%M:%S')} - {user} {action} '{title}'")

# Example Usage:
library = Library("Central Library")

# Create librarian
librarian = Librarian("Alice", "alice@library.com", "L001")

# Create books
book1 = Book("1984", "George Orwell", "ISBN001")
book2 = Book("To Kill a Mockingbird", "Harper Lee", "ISBN002")
book3 = Book("The Great Gatsby", "F. Scott Fitzgerald", "ISBN003")

# Librarian adds books
librarian.add_book(book1, library)
librarian.add_book(book2, library)
librarian.add_book(book3, library)

# Display current books
library.display_books()

# Create members
member1 = Member("Bob", "bob@example.com", "M001")
member2 = Member("Carol", "carol@example.com", "M002")

# Members borrow books
member1.borrow_book(book1, library)
member2.borrow_book(book2, library)

# Try to borrow a book already borrowed
member1.borrow_book(book2, library)

# Display current books
library.display_books()

# Return books
member1.return_book(book1, library)

# Display history
library.show_history()
```
